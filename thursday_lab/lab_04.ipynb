```python
import pandas as pd
import numpy as np
import scipy.stats as stats
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# ECDF plot function with default random number generator for reproducibility
rng = np.random.default_rng(0)
def plot_ecdf(sample, theoretical_cdf, dist_name):
    ax = sns.ecdfplot(sample, label = 'ECDF')
    x_grid = np.linspace(*ax.get_xlim(), 500)
    ax.plot(x_grid, theoretical_cdf(x_grid), 
            label = f"{dist_name} CDF (theory)", linestyle = "--")
    ax.set_xlabel("x")
    ax.legend()
    plt.show()
```

```python
# Q1 — discrete grid warm-up: approaches Uniform(0, 1) as K grows
K = 100_000  # try shrinking this (e.g. 10, 100) to see the ECDF get "steppy"
n = 5_000

grid = np.arange(1, K + 1) / K
draws_q1 = rng.choice(grid, size=n)

plot_ecdf(draws_q1, stats.uniform(loc=0, scale=1).cdf, dist_name="Uniform(0,1)")
```

```python
# Q2 — rescaled sample mean of Uniform draws: approaches Normal(0, 1) by the CLT
n = 32
b = 5_000
a = np.sqrt(3)

samples_q2 = rng.uniform(-a, a, size=(b, n))
means_q2 = samples_q2.mean(axis=1) * np.sqrt(n)

plot_ecdf(means_q2, stats.norm(loc=0, scale=1).cdf, dist_name="Normal(0,1)")
```

```python
# Q3 — standardized coin-flip counts: approaches Normal(0, 1) by the CLT
n = 40
p = 0.5  # try 0.1 or 0.01 to see the Normal approximation get worse
b = 5_000

heads = rng.binomial(n, p, size=b)
standardized_q3 = (heads - n * p) / np.sqrt(n * p * (1 - p))

plot_ecdf(standardized_q3, stats.norm(loc=0, scale=1).cdf, dist_name="Normal(0,1)")
```

```python
# Q4 — discrete-time survival process: approaches Exponential(rate=r) as dt -> 0
dt = 1e-3  # try 1e-1 or 1e-2 to see the Exponential approximation get worse
r = 2.0    # try changing this to see the rate of the limiting Exponential change
n = 5_000

periods = rng.geometric(p=r * dt, size=n)  # period in which termination occurs
times_q4 = periods * dt

plot_ecdf(times_q4, stats.expon(scale=1 / r).cdf, dist_name="Exponential")
```
